```python
import pandas as pd
df=pd.read_csv('Dataset.csv')

#1. Top cuisines analysis#
d=df.dropna(subset=['Cuisines'])
c=d['Cuisines'].str.split(',').explode().value_counts().head(3)
tot=len(d)
for name,cnt in c.items():
    pct=(cnt/tot)*100
    print(f"{name}:{cnt} ({pct: .2f}%)")
```

```python
#2.City wise restaurant insights#
top_city= df['City'].value_counts().index[0]
top_cnt= df['City'].value_counts().iloc[0]
print(f"topcity:{top_city} ({ top_cnt}")

avg =df.groupby('City')['Aggregate rating'].mean()
print(f" best rating city is {avg.idxmax()} ({avg.max(): .2f})")
```

```python
#3. Price range distribution#
p=df['Price range'].value_counts()
tot=len(df)
for rng,cnt in p.items():
    pct=round((cnt/tot)*100,2)
    print("price range" ,rng, ":",cnt,pct,"%")
```

```python
#4.Online delivery availability#
deli=df['Has Online delivery'].value_counts(normalize=True)*100
book= df['Has Table booking'].value_counts(normalize=True)*100

print('online delivery%:')
print(deli.round(2))
print('table booking%:')
print(book.round(2))
avg_deli=df.groupby('Has Online delivery')['Aggregate rating'].mean()
print("avg rating( no delivery VS delivery is")
print(avg_deli.round(2))
```

```python
#5.Table booking & price range distribution#
tb=df['Has Table booking'].value_counts(normalize=True)*100
od=df['Has Online delivery'].value_counts(normalize=True)*100
print("table booking %:")
print(tb.round(2))
print('Online Delivery % :')
print(od.round(2))

tb_price=df.groupby('Price range')['Has Table booking'].value_counts(normalize=True)*100
print("table booking by price range %:")
print(tb_price.round(2))
```

```python
#6. Price tier vs rating performance#
com= df['Price range'].mode()[0]
print("most common price range:" , com)
av= df.groupby('Price range')['Aggregate rating'].mean()
print(' avg rating by price range:')
print(av.round(2))
```

```python
#7. Restaurant name length & feature combination#
df['Name Length']=df['Restaurant Name'].apply(len)
df['Has Both']=((df['Has Table booking']=='Yes')&(df['Has Online delivery']=='Yes')).astype(int)
print('name length head:')
print(df['Name Length'].head())
print("has both count")
print(df['Has Both'].value_counts())
```

```python
#8. Customer votes by rating category#
v= df.groupby('Rating color')['Votes'].sum()
print('total votes by rating color:')
print(v)
```
